## 3. Download del dataset (Telco Customer Churn)
Il dataset viene scaricato **sul server di Colab**. Non serve un account Kaggle.

```python
import os

try:
    path = kagglehub.dataset_download("blastchar/telco-customer-churn")
    df = pd.read_csv(os.path.join(path, "WA_Fn-UseC_-Telco-Customer-Churn.csv"))
    fonte = "Kaggle"
except Exception as e:
    print("⚠️ Kaggle non raggiungibile:", e)
    print("   Provo la copia su OpenML...")
    from sklearn.datasets import fetch_openml
    df = fetch_openml(data_id=42178, as_frame=True).frame
    fonte = "OpenML"

print(f"Fonte: {fonte}  |  Righe: {df.shape[0]}  |  Colonne: {df.shape[1]}")
assert df.shape[0] == 7043, "Numero di righe inatteso"
print("\n✅ Dataset caricato")
df.head()
```

## 4. Un grafico veloce

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))

df["Churn"].value_counts().plot(kind="bar", ax=ax[0], color=["#4C72B0", "#DD8452"])
ax[0].set_title("Clienti che hanno lasciato (Churn)")
ax[0].set_xlabel("")

ax[1].hist(df["tenure"].astype(float), bins=30, color="#4C72B0")
ax[1].set_title("Mesi come cliente (tenure)")

plt.tight_layout()
plt.show()
print("✅ Grafici OK")
```
